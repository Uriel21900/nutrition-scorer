# Reproducing Baseline Nutritional Scoring

In this notebook, I will reproduce the standard mathematical algorithm used by the European "Nutri-Score" system using the open-source `pyNutriScore` package. 

The goal is to calculate the official baseline score for a few benchmark foods used in my Capstone project, and demonstrate where the standard algorithm fails (e.g., heavily penalizing healthy fats like Olive Oil, while rewarding artificial diet drinks).

```python
!pip install pynutriscore
```

```python
from nutriscore import NutriScore

# Initialize the calculator for standard solid foods and beverages
nutriscore = NutriScore()
```

### Test Case 1: Extra Virgin Olive Oil (Graza)
A pure, healthy fat. My custom Capstone app scores this a **9.5 / 10** because it identifies "extra virgin olive oil" as a heart-healthy superfood with zero glycemic impact.

Let's see what the standard algorithmic baseline gives it (Values per 100g):

```python
# Macros for 100g of Extra Virgin Olive Oil
evoo_nutrition = {
    'energy': 3389,             # kJ (~810 kcal)
    'sugars': 0,
    'saturated_fat': 14,
    'sodium': 0,
    'fruits_vegetables_nuts': 0,
    'fiber': 0,
    'proteins': 0
}

# Note: NutriScore has a specific 'fats' category, but we will test it under the standard rule to see the raw output.
score = nutriscore.calculate(evoo_nutrition)
grade = nutriscore.calculate_class(evoo_nutrition)

print(f"Baseline Nutri-Score Value: {score}")
print(f"Baseline Letter Grade: {grade}")
```

**Analysis:** The baseline model typically grades Olive Oil as a "C" or "D" because it blindly penalizes calorie density and saturated fat, without text-parsing the ingredient list to realize it is a premium healthy fat.

---
### Test Case 2: Crystal Light Diet Drink
A highly processed beverage full of artificial sweeteners (aspartame). My custom app scores this a **3.5 / 10** due to heavy penalties for artificial ingredients.

```python
# Macros for 100ml of Diet Drink
diet_drink_nutrition = {
    'energy': 5,              # kJ (~1 kcal)
    'sugars': 0,
    'saturated_fat': 0,
    'sodium': 10,
    'fruits_vegetables_nuts': 0,
    'fiber': 0,
    'proteins': 0
}

score = nutriscore.calculate(diet_drink_nutrition, beverage=True)
grade = nutriscore.calculate_class(diet_drink_nutrition, beverage=True)

print(f"Baseline Nutri-Score Value: {score}")
print(f"Baseline Letter Grade: {grade}")
```

**Analysis:** The baseline model awards this a "B" (sometimes an "A" depending on the exact calculator version) because it only sees zero calories and zero sugar. It cannot read the ingredient list to penalize the artificial chemical additives.
